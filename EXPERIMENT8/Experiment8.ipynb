```python
import cv2
import numpy as np
```

```python
img = cv2.imread('/content/WhatsApp Image 2026-03-23 at 9.13.53 AM.jpeg')
```

```python
gray =cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
```

```python
hsv = cv2.cvtColor(img,cv2.COLOR_BGR2HSV)
```

```python
h,s,v = cv2.split(hsv)
```

```python
lower_h=35
upper_h=85
```

```python
mask=cv2.inRange(h,lower_h,upper_h)
```

```python
result=cv2.bitwise_and(img,img,mask=mask)
```

```python
cv2.imwrite("green.jpg",result)
```
